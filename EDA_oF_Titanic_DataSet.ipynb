```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv(r'C:\Users\hyper\OneDrive\Desktop\Internship\Task5\Titanic_Data_set.csv',encoding = 'unicode_escape')
```

```python
# Now Replacing Null values of cabin with 'U' means 'Unknown'.
df.fillna('U',inplace=True)
```

```python
# Converting age into year which is in year+decimal.
df['Age'] = df['Age'].astype(int)
```

```python
df.columns
```

```python
# Now We Creating EDA for this Cleaned Data.
ax = sns.countplot(x='Survived', data=df , palette='Set2')
```

```python
# Count according to class.
ax=sns.countplot(x='Pclass', data=df, palette='Set1')
```

```python
# count according to sex distribution.
ax = sns.countplot(x='Sex', data=df, palette='Set3')
```

```python
# Count of age.
column_data = df['Age']
plt.hist(column_data,bins=20,edgecolor='Black')
```

```python
# Count of Fare
column_data=df['Fare']
plt.hist(column_data,bins=20,edgecolor='Black')
```

```python
# count of Embarked.
sns.countplot(x='Embarked', hue='Survived', data=df)
```

```python
# survived people according to Pclass.
ax = sns.barplot(x='Pclass',y='Survived',data=df,palette='Set1')
```

```python
# Survived people according to gender.
ax= sns.barplot(x='Sex',y='Survived', data=df , palette='Set2')
```

```python
# Survived by Embarked.
ax= sns.barplot(x='Embarked',y='Survived', data=df , palette='Set2')
```

```python
# Survived by Age.
ax=sns.violinplot(x='Survived', y='Age',data=df)
```

```python
# Survivbved by Fare.
sns.boxplot(x='Survived', y='Fare', data=df)
```

```python
# Combined Plotting.
sns.catplot(x='Pclass', hue='Sex', col='Survived', kind='count', data=df)
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
```
